```python


import pandas as pd 
movie=pd.read_csv(r"C:\Users\Dev\Desktop\pillow\data\tmdb_5000_movies.csv", encoding="latin1")

credits = pd.read_csv(
    r"C:\Users\Dev\Desktop\pillow\data\tmdb_5000_credits.csv"
)
```

```python
movies = movies[
    [
        "id",
        "title",
        "overview",
        "genres",
        "keywords",
        "cast",
        "crew",
        "popularity",
        "vote_average",
        "vote_count"
    ]
]
print(movies.head())
```

```python
print(movies.info())
```

```python
import ast
def convert(text):
    result=[]
    for item in ast.literal_eval(text):
        result.append(item["name"])
    return result
```

```python
movies["genres"] = movies["genres"].apply(convert)
movies["keywords"] = movies["keywords"].apply(convert)
movies["cast"] = movies["cast"].apply(convert)
movies["crew"] = movies["crew"].apply(convert)
movies["cast"].iloc[0]
```
